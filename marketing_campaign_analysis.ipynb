```python
import pandas as pd
import os

# Go one level up from notebooks folder
project_path = os.path.dirname(os.getcwd())

data_path = os.path.join(project_path, "data")


nykaa = pd.read_csv(
    os.path.join(data_path, "nykaa_campaign_data_with_nulls.csv")
)

purplle = pd.read_csv(
    os.path.join(data_path, "purplle_campaign_data_with_nulls.csv")
)

tira = pd.read_csv(
    os.path.join(data_path, "tira_campaign_data_with_nulls.csv")
)


print("Nykaa Shape:", nykaa.shape)
print("Purplle Shape:", purplle.shape)
print("Tira Shape:", tira.shape)
```

```python
nykaa["Brand"] = "Nykaa"

purplle["Brand"] = "Purplle"

tira["Brand"] = "Tira"
```

```python
marketing_df = pd.concat(
    [
        nykaa,
        purplle,
        tira
    ],
    ignore_index=True
)
```

```python
missing_percentage = (
    marketing_df.isnull().sum() / len(marketing_df)
) * 100

missing_percentage.sort_values(
    ascending=False
)
```

```python
numeric_columns = marketing_df.select_dtypes(
    include=['float64']
).columns


for col in numeric_columns:
    marketing_df[col] = marketing_df[col].fillna(
        marketing_df[col].median()
    )
```

```python
categorical_columns = marketing_df.select_dtypes(
    include=['object', 'str']
).columns


for col in categorical_columns:
    marketing_df[col] = marketing_df[col].fillna(
        marketing_df[col].mode()[0]
    )
```

```python
duplicate_count = marketing_df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
marketing_df["Date"] = pd.to_datetime(
    marketing_df["Date"],
    dayfirst=True
)
```

```python
marketing_df["Profit_Flag"] = marketing_df["ROI"].apply(
    lambda x: 1 if x > 1 else 0
)
```

```python
from sklearn.preprocessing import MultiLabelBinarizer
```

```python
marketing_df["Channel_Used_List"] = (
    marketing_df["Channel_Used"]
    .str.split(", ")
)
```

```python
mlb = MultiLabelBinarizer()

channel_encoded = mlb.fit_transform(
    marketing_df["Channel_Used_List"]
)
```

```python
channel_df = pd.DataFrame(
    channel_encoded,
    columns=mlb.classes_,
    index=marketing_df.index
)
```

```python
channel_df.head()
```

```python
marketing_df = pd.concat(
    [
        marketing_df,
        channel_df
    ],
    axis=1
)
```

```python
marketing_df.drop(
    "Channel_Used_List",
    axis=1,
    inplace=True
)
```

```python
marketing_df.shape
```

```python
marketing_df.columns.tolist()
```

```python
marketing_df.dtypes
```

```python
regression_features = [
    "Duration",
    "Impressions",
    "Clicks",
    "Leads",
    "Conversions",
    "Acquisition_Cost",
    "ROI",
    "Engagement_Score",
    "Year",
    "Month",
    "Email",
    "Facebook",
    "Google",
    "Instagram",
    "WhatsApp",
    "YouTube"
]


X_reg = marketing_df[regression_features]

y_reg = marketing_df["Revenue"]
```

```python
X_reg.shape
y_reg.shape
```

```python
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr_model = LinearRegression()
```

```python
lr_model.fit(
    X_train,
    y_train
)
```

```python
lr_prediction = lr_model.predict(
    X_test
)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np


# MSE
mse = mean_squared_error(
    y_test,
    lr_prediction
)

# RMSE
rmse = np.sqrt(mse)

# MAE
mae = mean_absolute_error(
    y_test,
    lr_prediction
)

# R2 Score
r2 = r2_score(
    y_test,
    lr_prediction
)


print("RMSE:", rmse)
print("MAE:", mae)
print("R2 Score:", r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
```

```python
rf_model.fit(
    X_train,
    y_train
)
```

```python
rf_prediction = rf_model.predict(
    X_test
)
```

```python
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score


rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_prediction
    )
)

rf_mae = mean_absolute_error(
    y_test,
    rf_prediction
)

rf_r2 = r2_score(
    y_test,
    rf_prediction
)


print("Random Forest RMSE:", rf_rmse)
print("Random Forest MAE:", rf_mae)
print("Random Forest R2 Score:", rf_r2)
```

```python
import joblib

joblib.dump(
    rf_model,
    "revenue_prediction_model.pkl"
)
```

```python
classification_features = [
    "Duration",
    "Impressions",
    "Clicks",
    "Leads",
    "Conversions",
    "Acquisition_Cost",
    "Engagement_Score",
    "Year",
    "Month",
    "Email",
    "Facebook",
    "Google",
    "Instagram",
    "WhatsApp",
    "YouTube"
]


X_cls = marketing_df[classification_features]

y_cls = marketing_df["Profit_Flag"]
```

```python
print(X_cls.shape)
print(y_cls.shape)
```

```python
from sklearn.model_selection import train_test_split


X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls,
    y_cls,
    test_size=0.2,
    random_state=42,
    stratify=y_cls
)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
log_model = LogisticRegression(
    max_iter=1000
)
```

```python
log_model.fit(
    X_train_cls,
    y_train_cls
)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(
    X_train_cls
)

X_test_scaled = scaler.transform(
    X_test_cls
)
```

```python
from sklearn.linear_model import LogisticRegression


log_model = LogisticRegression(
    max_iter=2000
)


log_model.fit(
    X_train_scaled,
    y_train_cls
)
```

```python
log_prediction = log_model.predict(
    X_test_scaled
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


accuracy = accuracy_score(
    y_test_cls,
    log_prediction
)

precision = precision_score(
    y_test_cls,
    log_prediction
)

recall = recall_score(
    y_test_cls,
    log_prediction
)

f1 = f1_score(
    y_test_cls,
    log_prediction
)


print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

```python
from sklearn.tree import DecisionTreeClassifier


dt_model = DecisionTreeClassifier(
    random_state=42
)


dt_model.fit(
    X_train_cls,
    y_train_cls
)
```

```python
dt_prediction = dt_model.predict(
    X_test_cls
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


dt_accuracy = accuracy_score(
    y_test_cls,
    dt_prediction
)

dt_precision = precision_score(
    y_test_cls,
    dt_prediction
)

dt_recall = recall_score(
    y_test_cls,
    dt_prediction
)

dt_f1 = f1_score(
    y_test_cls,
    dt_prediction
)


print("Accuracy:", dt_accuracy)
print("Precision:", dt_precision)
print("Recall:", dt_recall)
print("F1 Score:", dt_f1)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_classifier = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
```

```python
rf_classifier.fit(
    X_train_cls,
    y_train_cls
)
```

```python
rf_cls_prediction = rf_classifier.predict(
    X_test_cls
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


rf_accuracy = accuracy_score(
    y_test_cls,
    rf_cls_prediction
)

rf_precision = precision_score(
    y_test_cls,
    rf_cls_prediction
)

rf_recall = recall_score(
    y_test_cls,
    rf_cls_prediction
)

rf_f1 = f1_score(
    y_test_cls,
    rf_cls_prediction
)


print("Accuracy:", rf_accuracy)
print("Precision:", rf_precision)
print("Recall:", rf_recall)
print("F1 Score:", rf_f1)
```
